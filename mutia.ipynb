```python
import math

print("KALKULATOR LINGKARAN")
r = float(input("Masukkan jari-jari lingkaran (r): "))

luas = math.pi * r**2
keliling = 2 * math.pi * r

print(f"Luas Lingkaran     : {luas:.2f}")
print(f"Keliling Lingkaran : {keliling:.2f}")
```

```python
print("KALKULATOR PERSEGI")
s = float(input("Masukkan panjang sisi persegi (s): "))

luas = s * s
keliling = 4 * s

print(f"Luas Persegi     : {luas:.2f}")
print(f"Keliling Persegi : {keliling:.2f}")
```

```python
print("KALKULATOR PERSEGI PANJANG")
p = float(input("Masukkan panjang (p): "))
l = float(input("Masukkan lebar (l): "))

luas = p * l
keliling = 2 * (p + l)

print(f"Luas Persegi Panjang     : {luas:.2f}")
print(f"Keliling Persegi Panjang : {keliling:.2f}")
```
